```python
from numpy import average
import pandas as pd
import seaborn as sns
df = sns.load_dataset("titanic")
print("=" * 60)
print(" Titanic Dataset Loaded Successfully ")
print("=" * 60)
print("\n first 5 rows : ")
print(df.head())
print("\n first 5 colomns : ")
print(df.tail())
print("\n dataset shape : ")
print("row : ",df.shape[0])
print("colomn : ",df.shape[1])
print("\n colomn name : ")
print(df.columns.tolist())
print("dataset information : ")
print(df.info())
print("numerical summary : ")
print(df.describe())
print("summary categorical colomns : ")
print(df.describe(include = "all"))
print("missing value count : ")
print(df.isnull().sum())
print("missing value percentage : ")
per = df.isnull().mean()*100
print(per.round(2))
print("duplicate rows : ")
print(df.duplicated().sum())
df_clean = df.copy()
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())
df_clean["embarked"] = df_clean["embarked"].fillna(df_clean["embarked"].mode()[0])
df_clean["embark_town"] = df_clean["embark_town"].fillna(df_clean["embark_town"].mode()[0])
df_clean["deck"] = df_clean["deck"].cat.add_categories(["unknown"]).fillna("unknown")
print("\n missing values after selected cleaning : ")
print(df_clean.isnull().sum())
print("\n duplicate check after cleaning : ")
print(df_clean.duplicated().sum())
print("\n passenger count gender wise : ")
print(df_clean["sex"].value_counts())
print("\n passenger count class wise : ")
print(df_clean["pclass"].value_counts().sort_index())
print("\n  survival count : ")
print(df_clean["survived"].value_counts().sort_index())
passenger = len(df_clean)
survivors = int(df_clean["survived"].sum())
not_survivors = int((df_clean["survived"]==0).sum())
rate = df_clean["survived"].mean()*100
print("  overall survival analysis : ")
print("total passenger : ", passenger)
print("survived : ",survivors)
print("not survived : ",not_survivors)
print("srvival rate : ",round(rate,2),"%")
print("  survival rate by gender : ")
gen = df_clean.groupby("sex").agg(passenger_count=("survived","count"),survivors=("survived","sum"),
                                  survivors_rate = ("survived","mean"))
gen["survivors_rate"] *= 100
print(gen.round(2))
print("survival rate by passenger class wise : ")
cl = df_clean.groupby("pclass").agg(passenger_count=("survived","count"),survivors=("survived","sum"),
                                  survivors_rate = ("survived","mean"),average_fare =("fare","mean"))
cl["survivors_rate"]*=100
print(cl.round(2))
print(" survival rate by class and gender wise : ")
cl_gen = df_clean.groupby(["pclass","sex"],observed=True).agg(passenger_count=("survived","count"),survivors=("survived","sum"),
                                  survivors_rate = ("survived","mean"))
cl_gen["survivors_rate"]*=100
print(cl_gen.round(2))
print(" age analysis by passenger class basis : ")
age_ana = df_clean.groupby("pclass").agg(average_age = ("age","mean"),age_max = ("age","max"),
                                         age_min = ("age","min"),passenger_count = ("age","count"))
print(age_ana.round(2))
df_clean["family_size"] = (df_clean["sibsp"]+df_clean["parch"]+1)
print(" survival rate by family size : ")
fam_ana = df_clean.groupby("family_size").agg(passenger_count = ("survived","count"),survivors = ("survived","sum"),
                                             survivors_rate = ("survived","mean"))
fam_ana["survivors_rate"]*=100
print(fam_ana.round(2))
```
